**Aprendizado Superviosionado X Não Supervisionado**

Supervisionado:     [ Dados (X) ] + [ Rótulos (Y) ]  ---> Modelo aprende a prever Y

Não Supervisionado: [ Dados (X) ]                   ---> Modelo descobre a estrutura

**Objetivo do Aprendizado Não Supervisionado**

Encontrar padrões ocultos dentro dos próprios dados, mas sem um gabarito(rótulo)

Os modelos aprendezem sozinhos a descobrir coisas utilizando algorítimos\
para diversas situações:

- Simplicar conjuntos de dados gigantes e acelerar resultados (Utilizado em visãoc computacional)
- Uso de analogias para detecar padrões, detecção de anomalias
- Realizar agrupamentos

**Conceito Matemático fundamental**
Existe uma distância entre os "pontos de dados" que permite dizer quão parecidos são esses pontos?

Analiticamente existem 2 métodos de cálculode de distancia para responder essas perguntas, e os algoritmos são baseados nesses fundamentos.


- Distancia euclidiana: É basicamente a linha reta entre 2 pontos
- Distancia Manhattan: É a distancia percorrida em 2D ou seja em angulos de 90º  |__
